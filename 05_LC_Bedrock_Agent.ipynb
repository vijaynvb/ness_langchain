Agents act as intermediaries that can dynamically select and execute actions based on the input they receive. They enable the creation of complex applications that require decision-making and multi-step processes.

TavilySearchResults is a valuable tool that connects your language model (LLM) to the web using the Tavily Search API.

```python
from langchain.agents import create_agent
from langchain_aws import ChatBedrock
from langchain_core.messages import AIMessage, HumanMessage
from langchain_core.outputs import ChatGeneration, ChatResult
from langchain_core.runnables import Runnable
from pydantic import ConfigDict
```

## Prompt

**LangChain Hub**

Efficiently manage your LLM components with the LangChain Hub.

## LLM

```python
# change the model_id to the one you want to use
# For example, you can use "claude-3.5-sonnet" or any other model which supports function calling
# You can find the available models in the AWS Bedrock console or documentation.
llm = ChatBedrock(
    model_id="mistral.mistral-7b-instruct-v0:2",  # Replace with your model ID
    temperature=0,
)
```

## Tools

Tavily's Search API is a search engine built specifically for AI agents (LLMs), delivering real-time, accurate, and factual results at speed.

Generate Tavily API Key from this [link](https://app.tavily.com/home)

```python
import os
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

os.environ["TAVILY_API_KEY"] = os.getenv('TAVILY_API_KEY')
```

```python
from langchain_community.tools.tavily_search import TavilySearchResults as TavilySearch

search = TavilySearch()
tools = [search]
```

```python
prompt = (
    "You are a helpful assistant. Use the Tavily search tool for current information "
    "and answer clearly and concisely."
    "what's the weather in Bengaluru?"
)
prompt
```

## Agent

```python
agent = create_agent(model=llm, tools=tools, system_prompt=prompt)
```

```python
result = agent.invoke({"input": "what's the weather in Bengaluru?"})
```
